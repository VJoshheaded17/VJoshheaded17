# Retrieval: lexical, MiniLM, and CLIP

The first cell runs offline on preserved, unverified model captions. Neural encoding and vector storage are opt-in and download models. CLIP image retrieval needs original images; the historical FAISS index held only caption vectors.


```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT / "src"))
from georag.data import read_jsonl
from georag.retrieval import LexicalRetriever
documents = read_jsonl(ROOT / "data/sample/legacy_captions.jsonl")
retriever = LexicalRetriever(documents)
retriever.search("urban area with a river", top_k=3)
```

```python
RUN_NEURAL = False
if RUN_NEURAL:
    from georag.embeddings import MiniLMEncoder
    from georag.vector_store import VectorRetriever
    neural = VectorRetriever(documents, MiniLMEncoder(), backend="qdrant")
    try:
        print(neural.search("urban area with a river", top_k=3))
    finally:
        neural.close()
```
