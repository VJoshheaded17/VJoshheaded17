# Paired RAG evaluation

Preview both conditions without contacting an API. Each receives the same question, target identifier, system instructions and generation settings. RAG adds retrieved captions. Text overlap is not factual accuracy, and RAG may propagate incorrect captions. The original stored BLEU values are not a validated benchmark.


```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT / "src"))
from georag.prompts import answer_messages
from georag.evaluation import retrieval_metrics, claim_metrics
question = "Describe the broad land-cover features of the target scene."
target = {"scene_id": "your-scene-001"}
context = [{"id": "illustration", "caption": "A river crosses built-up land.", "verified": False}]
no_rag = answer_messages(question, target=target)
rag = answer_messages(question, context, target)
print(no_rag)
print(rag)
```

```python
print(retrieval_metrics(["a", "b", "c"], ["b"], k=3))
# Illustrative manual labels, not annotations of the original imagery:
print(claim_metrics(["supported", "unsupported", "unverifiable"]))
```
