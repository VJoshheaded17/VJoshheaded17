# Kosmos-2 baseline

Derived from the original independent-study notebooks. This first prompt preview runs without model downloads. The inference cell is opt-in: provide an actual image and set `RUN_INFERENCE = True`. No satellite pixels were included in the uploaded files.


```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT / "src"))
from georag.prompts import caption_prompt
metadata = {"representation": "RGB"}
print(caption_prompt(metadata, mode="image_only"))
```

```python
RUN_INFERENCE = False
image_path = ROOT / "data/raw/your_image.png"
if RUN_INFERENCE:
    from georag.caption import KosmosCaptioner
    model = KosmosCaptioner()
    result = model.describe(image_path, metadata, mode="image_only")
    print(result["caption"])
```
